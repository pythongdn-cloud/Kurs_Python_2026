```python
import pandas as pd
```

```python
temperatura = pd.Series([10,-9,8,30])
print(temperatura)
```

```python
temperatura = pd.Series([10, -9, 8, 30], index = ['pon', 'wt', 'sr', 'sr'])
print(temperatura)
```

```python
temperatura = pd.Series({
    "pon": 10,
    "wt": -9,
    "sr": 8,
    "sr": 30
})
print(temperatura)
```

```python
temperatura = pd.Series({
    "pon": 10,
    "wt": -9,
    "sr": 8,
    "sr": 30
}, name = "week1")
print(temperatura)
```
