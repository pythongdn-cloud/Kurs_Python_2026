# Metoda melt

```python
dane = {
    'Miasto' : ['Warszawa', 'Gdańsk'],
    'Data' : ['29 OCT', '30 OCT'],
    'Temperatura_rano' : [3, 5],
    'Temperatura_wieczorem' : [14, 11]
}
df = pd.DataFrame(dane)
df
```

```python
df_long = df.melt(
    id_vars = ['Miasto', 'Data'],
    value_vars = ['Temperatura_rano', 'Temperatura_wieczorem'],
    var_name = 'Pora dnia',
    value_name = 'Temperatura'
)
print(df_long) #dane w formacie long
```
